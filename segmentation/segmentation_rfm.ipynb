```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.mixture import GaussianMixture

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('segmentation_df.csv')
df.head()
```

```python
df2 = df.drop(columns = ['household_key', 'total_quantity', 'avg_quantity', 'avg_basket_size', 'unique_visits_days','coupon_dependecy_ratio'], axis = 1)
df2.head()
```

```python
scaler = StandardScaler()
df2 = pd.DataFrame(scaler.fit_transform(df2), columns = df2.columns)
df2.head()
```

```python
pca = PCA(n_components = 6)
principal_components = pca.fit_transform(df2)
principal_components
```

```python
eigen_values = pca.explained_variance_
eigen_values
```

```python
pca.explained_variance_ratio_
```

```python
cumulative_var = np.cumsum(eigen_values)
cumulative_var
```

- Using only the RFM matrix, the explained variance comes out nearly 59% which is low than the minimum ideal variance (70% to 90%)
