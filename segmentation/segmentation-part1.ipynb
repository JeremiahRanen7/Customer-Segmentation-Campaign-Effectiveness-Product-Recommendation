#### FEATURES TO BE CREATED

1. Features related to spending - sales_value,quantity
2. Features related to frequency - Basket_Id
3. Features related to recency - Day
4. Features related to discount - Coupon_disc...

```python
customer_spending = df2.groupby('household_key').agg({
    'SALES_VALUE':['sum','mean','min','max'],
    'QUANTITY':['sum','mean','min','max'],
})
```

```python
customer_spending.head()
```

```python
customer_spending.columns = ['total_sales','avg_sales','min_sales','max_sales','total_quantity','avg_quantity','min_quantity','max_quantity']

customer_spending.head()
```

```python
customer_spending.reset_index(inplace=True)
```

```python
customer_spending.head()
```

**FREQUENCY FEATURES**

```python
frequency_features = df2.groupby('household_key').agg({
    'BASKET_ID': ['nunique'],
    'DAY': ['nunique', 'min','max']
})
```

```python
frequency_features.columns = ['total_visits','unique_visits_days','first_purchase','last_purchase']
```

```python
frequency_features.reset_index(inplace=True)
```

```python
frequency_features.head()
```

```python
frequency_features['active_window'] =  (frequency_features['last_purchase'] - frequency_features['first_purchase'])

frequency_features['order_frequency'] = (frequency_features['active_window'] / frequency_features['total_visits'])
```

```python
customer_spending['avg_basket_size'] = customer_spending['total_quantity'] / frequency_features['total_visits']
```

```python
frequency_features.columns = ['household_key', 'total_orders','unique_visits_days','first_purchase','last_purchase', 'active_window', 'order_frequency']
```

```python
frequency_features.head()
```

**RECENCY FEATURES**

```python
# unique baskets per household
visits = df2.drop_duplicates(subset=['household_key', 'BASKET_ID'])[['household_key', 'BASKET_ID', 'DAY']]

# unique days per household
visit_days = visits.drop_duplicates(subset=['household_key', 'DAY'])[['household_key', 'DAY']]

# avg days between visits
recency_features = (visit_days.groupby('household_key')['DAY'].apply(lambda x: x.sort_values().diff().mean()).reset_index())
recency_features.columns = ['household_key', 'avg_days_between_visits']

# relative recency (last day of study is 712)
relative_recency = visit_days.groupby('household_key')['DAY'].agg(first_day='min', last_day='max').reset_index()
relative_recency['active_window'] = (relative_recency['last_day'] - relative_recency['first_day'])
relative_recency['relative_recency'] = (712 - relative_recency['last_day']).replace(0, 1)

relative_recency_clean = relative_recency[['household_key', 'relative_recency']]
```

```python
disc_features=[]
discount_features = pd.DataFrame()

total_sales = df2.groupby('household_key').agg({
    'SALES_VALUE': ['sum']
})

coupon_dependent_sales = df2[df2['COUPON_DISC'] != 0].groupby('household_key').agg({
    'SALES_VALUE': ['sum']
})

discount_features = pd.concat([discount_features,total_sales,coupon_dependent_sales], axis=1)
discount_features.columns = ['total_sales','coupon_dependent_sales']
discount_features['coupon_dependecy_ratio'] = discount_features['coupon_dependent_sales'] / discount_features['total_sales']
discount_features.reset_index(inplace=True)
discount_features.fillna(0, inplace=True)
discount_features.head()
```
